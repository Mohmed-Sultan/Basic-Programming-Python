```python
x = input("Enter password: ")

digit = False
uppercase = False

for char in x:
    if char.isdigit():
        digit = True

    if char.isupper():
        uppercase = True

if len(x) >= 8 and digit == True and uppercase == True:
    print("Valid Password")
else:
    print("Invalid Password")
```

```python
words = ["apple", "banana", "avocado", "blueberry", "cherry", "apricot"]

groups = {}

for w in words:
    if w[0] not in groups:
        groups[w[0]] = []

    groups[w[0]].append(w)

print(groups)
```

```python
cart = {
    "Laptop": 25000,
    "Mouse": 500,
    "Keyboard": 1200,
    "Headphones": 900
}

total = 0
count = 0

for product in cart:
    print(product, ":", cart[product])

    total += cart[product]

    if cart[product] > 1000:
        count += 1

print("Total:", total)
print("Products over 1000:", count)

if total > 25000:
    print("Expensive Order")
else:
    print("Normal Order")
```

```python
list1 = [1, 2, 3, 4, 5, 8]
list2 = [3, 4, 5, 6, 7, 8]

common = []

for x in list1:
    for y in list2:
        if x == y:
            common.append(x)

result = tuple(common)

print("Common values:", result)
```

```python
students = {
    "Ahmed": [70, 80, 90],
    "Sara": [95, 88, 92],
    "Omar": [40, 55, 45],
    "Mona": [60, 75, 70]
}

passed = []
failed = []

for name, grades in students.items():
    average = sum(grades) / len(grades)

    print(name, "Average:", average)

    if average >= 50:
        print("Passed")
        passed.append(name)
    else:
        print("Failed")
        failed.append(name)

print("Passed Students:", passed)
print("Failed Students:", failed)
```
